```python
import json
```

```python
def clean_data(data):
    # Remove users with missing names
    data["users"] = [user for user in data["users"] if user["name"].strip()]   # find empty space 

    # Remove dublicate friends
    for user in data["users"]:                             
        user['friends'] = list(set(user['friends']))      
    
    # Remove inactive users                                                                           
    data['users'] = [user for user in data['users']  if user['friends']   or  user['liked_pages']]   
                                                                                                      # Result Amit is removed

    # Remove dublicate pages        (it can solve in many ways,like.. set,dictionary)
    unique_pages = {}
    for page in data['pages']:                       
        unique_pages[page['id']] = page              # Result--> 104 update
    data['pages'] = list(unique_pages.values())      
                                                     # id 3 remove 
                                                     # page 4 remove
    return data
 

# Load the data
data = json.load(open("data2.json"))
data = clean_data(data)
json.dump(data, open("cleaned_data2.json", "w"), indent=4)
print("Data has been cleaned successfully")
```
